# Sentence Transformers to create vector embeddings

#### Note: requires Python >= 3.10

## Import packages.

```python
import numpy as np
import pandas as pd
from sentence_transformers import SentenceTransformer
```

## Read in csv file as a Pandas DataFrame and define the Sentence Transformer model.

```python
df = pd.read_csv("movies.csv")
model = SentenceTransformer("all-MiniLM-L6-v2") # Source: https://sbert.net/docs/sentence_transformer/usage/usage.html
```

## Get vector embeddings of the "overview" column.

```python
overviews = df["overview"]
```

```python
embeddings = []
for text in overviews:
    if type(text) == str:
        embeddings.append(model.encode(text)) # Result in a 1D array with 384 elements.
    else: # Keep track of nan values to be consistent with the rest of the original data.
        embeddings.append([text for i in range(len(embeddings[0]))])
```

```python
# 4803 rows and 384 columns.
embeddings_array = np.array(embeddings)
print(embeddings_array.shape)
```

## Convert to a DataFrame and csv file.

```python
vector_df = pd.DataFrame(embeddings_array)
vector_df
```

```python
vector_df.to_csv("Overview_Vector_Embeddings.csv")
```
